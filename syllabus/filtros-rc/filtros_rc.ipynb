# Filtros RC
> *Notas pedagógicas — S12 Curso de Filtros RC*

## 🧠 Punto de partida

El filtro RC es un divisor de voltaje donde **uno de los bloques cambia de tamaño según la frecuencia de la señal**.

```
Vin ──[aceite R]──● Vout ──[bloque elástico C]── tierra
```

- El bloque de aceite (resistor R) tiene longitud fija → caída de velocidad constante
- El bloque elástico (capacitor C) tiene **longitud efectiva que depende de la frecuencia**

→ Vout depende de cuánto del recorrido total ocupa cada bloque, y eso cambia con la frecuencia.

## 🎸 El capacitor como bloque elástico

La bola no pasa *a través* del bloque elástico — lo comprime.

**Frecuencia baja (señal lenta):**
- La bola empuja despacio → el bloque tiene tiempo de comprimirse hasta el fondo
- El bloque actúa como si fuera **muy largo** → absorbe casi toda la caída de velocidad
- Vout ≈ 0 → señal atenuada

**Frecuencia alta (señal rápida):**
- La bola llega y se va antes de que el bloque pueda comprimirse mucho
- El bloque actúa como si fuera **muy corto** → casi no afecta la velocidad
- Vout ≈ Vin → señal pasa completa

→ **Filtro pasa-altos**: señales rápidas pasan, señales lentas no.

*(Para filtro pasa-bajos: intercambiar R y C de lugar — el capacitor queda del lado de Vin)*

## 📐 La frecuencia de corte

Existe una frecuencia exacta donde el bloque elástico y el bloque de aceite tienen **la misma longitud efectiva** → divisor justo → Vout = Vin/2 (en términos de potencia: −3dB).

$$f_c = \frac{1}{2\pi RC}$$

- Por debajo de fc: el capacitor domina → señal atenuada
- Por encima de fc: el resistor domina → señal pasa

*(Esta ecuación puede introducirse después de que la intuición ya esté clara — no antes)*

## 🎸 Conexión con la perilla de tono

La perilla de tono de una guitarra eléctrica **ajusta el valor de C**:
- C grande → fc baja → más frecuencias graves pasan, agudos atenuados → sonido oscuro
- C pequeño → fc alta → más frecuencias agudas pasan → sonido brillante

La perilla *mueve* la frontera entre lo que pasa y lo que no.
