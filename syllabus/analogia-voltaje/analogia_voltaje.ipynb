# Analogía del Voltaje
> *Notas pedagógicas — S12 Curso de Filtros RC*

---

## 🧠 Insight central

**El voltaje es una propiedad del espacio, no del objeto.**

La confusión más común: el estudiante asocia voltaje con energía acumulada *en la carga*. Eso es el modelo de la mecánica clásica (una bola que rompe una pared con su energía cinética), pero no es cómo funciona el voltaje.

El voltaje es una propiedad del **campo eléctrico en ese punto del espacio**:
- No depende de qué carga pase por ahí
- No se gasta con el uso
- Es consistente: 5V → 0V siempre entrega la misma diferencia, sin importar la carga

## ❌ Modelo incorrecto (pero intuitivo)

> *"Una bola que quiere romper una pared"*

Si la bola tiene suficiente energía, rompe la pared y la atraviesa. Avanza un poco más cada vez por la energía residual.

**Por qué falla como analogía del voltaje:**
- La energía está *en el objeto* (la bola)
- El efecto depende de cuánta energía tenga el objeto
- La pared se modifica con cada interacción

Esto describe bien la *energía cinética*, no el potencial eléctrico.

## ✅ Modelo correcto — velocidad de la bola

> *"El voltaje es la velocidad de la bola en ese punto del espacio"*

**Reglas del modelo:**
- La bola viaja de la zona A (voltaje alto) a la zona B (voltaje bajo = tierra = 0)
- La **velocidad de la bola en cada punto = voltaje en ese punto**
- En la zona A la bola tiene velocidad máxima (Vin), en la zona B tiene velocidad 0
- La bola **siempre llega** de A a B — el voltaje no se agota, es una propiedad del espacio

**La resistencia = bloque de aceite:**
- A lo largo del bloque, la velocidad baja **linealmente** con la longitud
- Bloque más largo = más resistencia = caída de velocidad más pronunciada
- Al salir del último bloque la velocidad llega a 0 (tierra)

| Electricidad | Analogía |
|---|---|
| Voltaje en un punto (V) | Velocidad de la bola en ese punto |
| Voltaje de entrada (Vin) | Velocidad inicial de la bola |
| Resistencia (R) | Longitud del bloque de aceite |
| Tierra (0V) | Velocidad = 0 |

**Ventaja sobre la analogía de la montaña:** no requiere entender energía potencial — es directamente observable en la velocidad de un objeto físico.

## ⚡ Divisor de voltaje — dos bloques iguales

**Setup:** dos bloques de aceite de igual longitud en serie, de A a B.

```
A (Vin) ──[aceite L]──● nodo medio ──[aceite L]── B (0V)
```

**Lo que pasa:**
- La velocidad cae linealmente a lo largo de todo el recorrido
- Como los dos bloques son iguales (misma longitud L), el punto medio queda exactamente a la mitad del recorrido
- → En el nodo medio: **velocidad = Vin / 2**

**Generalización — bloques de distinto largo:**

```
A (Vin) ──[aceite R1]──● Vout ──[aceite R2]── B (0V)
```

La caída es proporcional a la longitud de cada bloque:

$$V_{out} = V_{in} \times \frac{R2}{R1 + R2}$$

**Por qué es intuitivo:** es pura interpolación lineal entre Vin y 0. No hace falta mencionar corriente — la geometría lo dice todo.

### Nota pedagógica
La corriente *podría* describirse como la velocidad inicial de la bola (más voltaje = bola más rápida = más corriente), pero **no es necesario para este curso**. El divisor de voltaje se entiende completamente con solo la caída de velocidad.
